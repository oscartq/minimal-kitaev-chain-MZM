## Run Simulation

Set parameters and run the simulation. The chemical potential will sweep from `-mu_max` to `+mu_max`. This may take several minutes depending on the system size.

```python
# Simulation parameters
nqbit = 2        # Number of qubits/orbitals
eps_max = 3.0     # Maximum chemical potential
nsteps = 20      # Number of steps in mu sweep
t = -1.0         # Hopping amplitude (must be real and negative)
Delta = 1.0      # Pairing amplitude (must be -t in topological regime)
const = 1.0      # Constant multiplier
eps = 0.0

# Run the simulation
results = run_simulation(nqbit=nqbit, eps_max=eps_max, nsteps=nsteps, 
                        t=t, Delta=Delta, const=const)
# 
# corr_results = compute_correlations(nqbit, eps, t=-1.0, Delta=1.0)

# # Save results
# with open('results/simulation_results.pkl', 'wb') as f:
#     pickle.dump(results, f)
# print("Results saved to 'simulation_results.pkl'")
# with open(f'results/corr_simulation_results_{eps}.pkl', 'wb') as f:
#     pickle.dump(corr_results, f)
# print("Results saved to 'simulation_results.pkl'")

# # Uncomment to load previously saved results
# with open('results/simulation_results.pkl', 'rb') as f:
#     results = pickle.load(f)
# print("Results loaded from 'simulation_results.pkl'")

with open(f'results/corr_simulation_results_{0.0}.pkl', 'rb') as f:
    corr_results = pickle.load(f)
print("Results loaded from 'corr_simulation_results.pkl'")
```
